# Part 1：線性轉換

## 1. 線性轉換的基本形式

線性轉換可以寫成：

$$
y = mx + b
$$

其中：

- $x$：原本的數值
- $y$：轉換後的數值
- $m$：斜率
- $b$：截距

例如，若想把：

```text
0  → 55
90 → 100
```

就可以找出適合的 $m$ 和 $b$。


## 2. 先取得最低分與最高分

Python 可以直接使用：

```python
min(scores)
max(scores)
```

例如：


````python
scores = [0, 20, 50, 70, 90]

lowest = min(scores)
highest = max(scores)

print("最低分：", lowest)
print("最高分：", highest)
print("範圍：", highest - lowest)
````

原本 notebook 中的函式：

```python
def find_ab(scores):
    a = min(scores)
    b = max(scores) - min(scores)
    return a, b
```

實際上回傳的是：

1. 最低分
2. 分數範圍（最高分 − 最低分）

因此改成更清楚的名稱會比較容易閱讀。


````python
def find_range_info(scores):
    lowest = min(scores)
    score_range = max(scores) - min(scores)

    return lowest, score_range


scores = list(map(int, input("請輸入數個分數：").split()))

lowest, score_range = find_range_info(scores)

print("最低分：", lowest)
print("分數範圍：", score_range)
````

## 3. 找出真正的 $m$ 與 $b$

假設要把原本的區間：

$$
[x_{\min}, x_{\max}]
$$

映射到：

$$
[y_{\min}, y_{\max}]
$$

則：

$$
m = \frac{y_{\max} - y_{\min}}
{x_{\max} - x_{\min}}
$$

接著：

$$
b = y_{\min} - mx_{\min}
$$

因此：

$$
y = mx + b
$$


````python
def find_linear_parameters(x_min, x_max, y_min, y_max):
    if x_max == x_min:
        raise ValueError("原始範圍不能只有一個數值")

    m = (y_max - y_min) / (x_max - x_min)
    b = y_min - m * x_min

    return m, b


m, b = find_linear_parameters(
    x_min=0,
    x_max=90,
    y_min=55,
    y_max=100
)

print("m =", m)
print("b =", b)
````

## 4. 使用線性轉換

建立：

```python
def transform(x, m, b):
    return m * x + b
```

就可以把原始分數轉換成新分數。


````python
def transform(x, m, b):
    return m * x + b


scores = [0, 20, 50, 70, 90]

m, b = find_linear_parameters(0, 90, 55, 100)

for score in scores:
    new_score = transform(score, m, b)
    print(score, "->", new_score)
````

### 和攝氏、華氏轉換的概念相同

攝氏溫度 $C$ 轉成華氏溫度 $F$：

$$
F = \frac{9}{5}C + 32
$$

也是：

$$
y = mx + b
$$

其中：

```text
m = 9 / 5
b = 32
```

例如：

```text
0°C   → 32°F
100°C → 212°F
```


````python
def celsius_to_fahrenheit(celsius):
    return celsius * 9 / 5 + 32


print(celsius_to_fahrenheit(0))
print(celsius_to_fahrenheit(100))
````

# Part 2：函式表示式解析

## 5. 三個函式

原本 notebook 定義了三個函式：


````python
def f(x):
    return 2 * x - 3


def g(x, y):
    return 2 * x + y - 7


def h(x, y, z):
    return 3 * x - 2 * y + z
````

三個函式需要的參數數量不同：

| 函式 | 參數數量 | 形式 |
|---|---:|---|
| `f` | 1 | `f(x)` |
| `g` | 2 | `g(x, y)` |
| `h` | 3 | `h(x, y, z)` |

例如：

```python
f(5)
g(3, 4)
h(7, 3, 3)
```


````python
print("f(5) =", f(5))
print("g(3, 4) =", g(3, 4))
print("h(7, 3, 3) =", h(7, 3, 3))
````

## 6. `split()` 將輸入切成 token

假設輸入：

```text
h f 5 g 3 4 3
```

使用：

```python
tokens = input().split()
```

會得到：

```python
["h", "f", "5", "g", "3", "4", "3"]
```

每一個項目可以稱為一個 **token**。


````python
tokens = "h f 5 g 3 4 3".split()

print(tokens)
````

## 7. 這個表示式代表什麼？

```text
h f 5 g 3 4 3
```

是一種 **前綴表示法（prefix notation）**。

依照每個函式需要的參數數量，可以還原成：

```python
h(
    f(5),
    g(3, 4),
    3
)
```

先算：

```python
f(5) = 7
g(3, 4) = 3
```

所以最後是：

```python
h(7, 3, 3)
```


````python
result = h(
    f(5),
    g(3, 4),
    3
)

print(result)
````

# 8. `pop(0)` 的作用

原本 notebook 有：

```python
lst = [1, 2, 3]

lst.pop(0)
```

`pop(0)` 會：

1. 取出索引 `0` 的元素
2. 同時把它從 list 中刪除

例如：


````python
lst = [1, 2, 3]

first = lst.pop(0)

print("取出的值：", first)
print("剩下：", lst)
````

再次執行：

```python
lst.pop(0)
```

就會再取出目前最前面的元素。


````python
lst = [1, 2, 3]

print(lst.pop(0))
print(lst)

print(lst.pop(0))
print(lst)
````

`pop(0)` 很適合用來理解「從左到右消耗 token」的概念。

不過在較大的資料中，每次從 list 最前面刪除元素都需要移動後面的資料。

所以正式解析時，可以改用：

- iterator
- index
- stack

下面使用 iterator 寫一個簡潔的遞迴版本。


# 9. 使用遞迴解析前綴表示式

基本想法：

1. 讀到數字 → 直接回傳數字
2. 讀到 `f` → 再讀 1 個表示式
3. 讀到 `g` → 再讀 2 個表示式
4. 讀到 `h` → 再讀 3 個表示式

因此一個函式裡可以再次呼叫自己。


````python
def calculate(tokens):
    tokens = iter(tokens)

    def parse():
        token = next(tokens)

        if token == "f":
            return f(parse())

        if token == "g":
            return g(parse(), parse())

        if token == "h":
            return h(parse(), parse(), parse())

        return int(token)

    result = parse()

    # 確認沒有多餘 token
    try:
        next(tokens)
        raise ValueError("輸入中有多餘的資料")
    except StopIteration:
        return result
````

## 10. 測試

輸入：

```text
h f 5 g 3 4 3
```

代表：

```python
h(f(5), g(3, 4), 3)
```


````python
tokens = "h f 5 g 3 4 3".split()

answer = calculate(tokens)

print(answer)
````

也可以處理巢狀函式，例如：

```text
f f f 2
```

代表：

```python
f(f(f(2)))
```


````python
tokens = "f f f 2".split()

print(calculate(tokens))
````

````python
text = input("請輸入表示式：").split()

print("結果：", calculate(text))
````

# 12. 執行流程範例

對：

```text
f f f 2
```

解析順序大致是：

```text
f
└─ f
   └─ f
      └─ 2
```

由最裡面開始計算：

```text
f(2)
→ f(f(2))
→ f(f(f(2)))
```

這就是遞迴的概念：

> 一個問題裡面包含相同形式的更小問題。
