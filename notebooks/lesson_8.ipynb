## 1. 圖形輸出概念

當輸入為 `3` 時，希望得到：

```text
__*
_**
***
```

當輸入為 `4` 時，希望得到：

```text
___*
__**
_***
****
```

可以觀察到：

- `_` 的數量逐行減少。
- `*` 的數量逐行增加。
- 若總共有 `n` 行，第 `i` 行可以由「底線 + 星號」組成。

## 3. 圖形輸出

### 原始學生版本

下面保留原本的寫法。要注意這一行：

```python
a = '_' * (n - 1), '*' * b
```

逗號會讓 `a` 變成 **tuple**，因此輸出會像：

```text
('__', '*')
('_', '**')
('', '***')
```

而不是直接得到預期的圖形。

````python
n = int(input())
b = 1

while n > 0:
    a = '_' * (n - 1), '*' * b
    print(a)
    n = n - 1
    b = b + 1
````

### 參考版本

若目標是把底線與星號直接接在一起，可以使用字串的 `+`：

````python
n = int(input())
stars = 1

while n > 0:
    line = '_' * (n - 1) + '*' * stars
    print(line)

    n -= 1
    stars += 1
````

## 4. 最大公因數：從較小的數往下找

原本的想法是：

以 `12` 與 `30` 為例，從較小的數 `12` 開始往下檢查：

```text
12, 11, 10, ..., 6, ..., 1
```

找到第一個可以同時整除兩個數的整數，就是最大公因數。

例如：

```text
12 % 6 == 0
30 % 6 == 0
```

所以：

```text
gcd(12, 30) = 6
```

### 原始學生版本

原程式先比較 `inputs[0]` 與 `inputs[1]`，但此時它們仍然是字串。

因此像 `"9"` 與 `"10"` 這類輸入，字串比較的結果可能與數值大小不同。這裡保留原版方便檢討。

````python
inputs = input().split()

if inputs[0] > inputs[1]:
    big = int(inputs[0])
    small = int(inputs[1])
else:
    big = int(inputs[1])
    small = int(inputs[0])

div = small

for i in range(div, 0, -1):
    if big % i == 0 and small % i == 0:
        print(i)
        break
````

### 較安全的參考版本

先把輸入轉成整數，再比較大小：

````python
a, b = map(int, input().split())

big = max(a, b)
small = min(a, b)

for divisor in range(small, 0, -1):
    if big % divisor == 0 and small % divisor == 0:
        print(divisor)
        break
````

## 5. 輾轉相除法（Euclidean Algorithm）

輾轉相除法可以更有效率地求最大公因數。

以 `754` 與 `221` 為例：

```text
754 % 221 = 91
221 % 91  = 39
91 % 39   = 13
39 % 13   = 0
```

當餘數變成 `0` 時，最後一個非零除數 `13` 就是最大公因數：

```text
gcd(754, 221) = 13
```

````python
a, b = map(int, input().split())

while b != 0:
    remainder = a % b
    a = b
    b = remainder

print(a)
````

### 更精簡的 Python 寫法

利用 Python 的多重指定，可以直接更新 `a` 與 `b`：

````python
a, b = map(int, input().split())

while b != 0:
    a, b = b, a % b

print(a)
````

## 6. `print()` 的回傳值是 `None`

`print()` 的功能是把內容輸出到畫面，但它本身**不會回傳有意義的資料**，因此回傳值為 `None`。

下面這段程式：

```python
a = print('a')
print(a)
```

第一行先輸出 `a`，並把 `print()` 的回傳值 `None` 存入變數 `a`。

所以最後會看到：

```text
a
None
```

````python
a = print('a')
print(a)
````
