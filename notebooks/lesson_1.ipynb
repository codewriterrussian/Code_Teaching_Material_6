## 1. 第一個輸出：`print()`

`print()` 可以把資料顯示在畫面上。

可以把程式想成：

```text
輸入 → 執行 → 輸出
```

下面的例子直接輸出已知字串 `hello world`。


````python
print('hello world')

# 輸入：已知的字串 hello world
# 執行：print()
# 輸出：hello world
````

### 一次輸出多個資料

`print()` 可以一次放入多個資料。


````python
print(123, '早安')
````

## 2. 使用 `sep` 控制資料之間的分隔符號

`sep` 是 separator（分隔符號）的縮寫。

預設情況下，`print()` 會使用空白分隔多個資料。


````python
print(123, '早安', sep=' ')
print(123, '早安', sep=',')
print(123, '早安', sep=' -> ')
````

### 更多 `sep` 範例


````python
print(1, 2, 3, 4, 5, sep=',')
print('台北', '台中', '高雄', sep=' -> ')
````

````python
print(1, 2, 3, 4, 5, sep='\\')
print(1, 2, 3, 4, 5, sep='^')
print(1, 2, 3, 4, 5, sep='&')
print(1, 2, 3, 4, 5, sep='$')
print(1, 2, 3, 4, 5, sep='#')
print(1, 2, 3, 4, 5, sep='~')
````

## 3. 換行與 `end`

每次執行 `print()` 時，預設會在最後加上換行字元 `\n`。


````python
print('h')  # h + \n
print('e')  # e + \n
print('')   # 空字串 + \n
````

### 使用 `end` 改變 `print()` 結尾

`end` 可以控制每次 `print()` 執行完成後要接什麼內容。


````python
print('h', end=',')
print('e', end=' ')
print('', end='\n')
````

### 練習：使用多次 `print()` 輸出同一行資料


````python
# 目標輸出：
# 1,2,3,4,5

print(1, end=',')
print(2, end=',')
print(3, end=',')
print(4, end=',')
print(5)
````

````python
# 目標輸出：
# 台北 -> 台中 -> 高雄

print('台北', end=' -> ')
print('台中', end=' -> ')
print('高雄')
````

## 4. 使用 `input()` 接收使用者輸入

`input()` 會等待使用者輸入資料。

如果希望把輸入內容顯示出來，可以把 `input()` 放進 `print()`。


````python
print(input())
````

### Notebook 與 `.py` 的差異

在 Jupyter Notebook 中，如果 `input()` 是 cell 的最後一行，Notebook 可能會直接顯示它回傳的結果。

但在一般 `.py` 程式中，`input()` 的主要作用是接收資料；如果要顯示資料，仍然需要使用 `print()`。

可以把它想成：

```text
食材放進烤箱 → input()
烤箱處理資料
把完成的東西拿出來 → print()
```


````python
input()
````

### 在 `input()` 中加入提示文字


````python
print(input('請輸入名字 年齡 城市：'))
````

## 5. 使用 `.split()` 拆分輸入

如果使用者一次輸入多個資料，可以使用 `.split()` 依照分隔符號拆開。


````python
name, age, city = input('請輸入名字 年齡 城市：').split()

print(name, age, city, sep='\n')
````

### 預設按照空白拆分

例如輸入：

```text
2 * ( 3 + 4 ) * 5
```

`.split()` 會依照空白把字串拆開。


````python
print(input('請輸入數學運算式：').split())
````

### 指定其他分隔符號

`.split(',')` 可以依照逗號拆分。


````python
print(input().split(','))
````

也可以使用較長的字串作為分隔符號，例如 ` -> `。


````python
# 範例輸入：
# 台北 -> 台中 -> 高雄

print(input().split(' -> '))
````

## 6. 基本變數操作

變數可以保存資料，也可以重新指定新的值。


````python
x = 10  # 定義變數 x 的值為 10

print(x + 10)

x = x + 10
print(x)
````

```text
年/月/日
```

**範例輸入**
