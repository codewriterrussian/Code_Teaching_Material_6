## 1. 跳脫字元

常見跳脫字元：
- `\\`：反斜線
- `\n`：換行
- `\t`：Tab


````python
print('A\nB')
print('\\')
print('A\tB\nC\tD\nE\tF\n')
````

## 2. `%` 字串格式化

常見格式：
- `%d`：整數
- `%f`：浮點數
- `%s`：字串


````python
print('%s %s %s' % (10, 20.2, ['a', 'b', 'c']))
````

````python
print('%d%s' % (20, 'Hi'))
````

### 欄寬與小數位數

`%8.2f`：
- `8`：欄位寬度
- `.2`：小數點後 2 位
- `f`：浮點數


````python
num = 5.2

print('%8.2f' % num)
````

## 3. `.format()` 字串格式化


````python
print('{}{}'.format(20, 'Hi'))
````

### 指定欄位索引


````python
print('{0}{1}{0}'.format('A', 'B'))
````

### 指定欄位寬度


````python
print('{0:3}{1:4}'.format(12, 345))

# 0 -> 索引
# :3 -> 欄位寬度
````

## 4. 練習：姓名與分數

輸入姓名和分數，輸出格式為「姓名 - 分數」，並另外輸出是否及格。

**範例輸入**
```text
Ben 85
```

**範例輸出**
```text
Ben - 85
True
```


````python
name, score = input().split()
score = int(score)

print('{} - {}'.format(name, score))
print(score >= 60)
````

## 5. f-string


````python
print(f'{name} - {score:.2f}')
````

## 7. 平均值格式化練習

回傳 `a` 和 `b` 的平均值。

原始筆記的輸出格式：
```text
average of (a) and (b) is (___)
```


````python
a, b = input().split()
a, b = int(a), int(b)
avg = (a + b) / 2

print('average of {} and {} is {:.2f}'.format(a, b, avg))
````
