### `and` 與 `or` 的優先順序

Python 會先處理 `and`，再處理 `or`。


````python
print(True or False and True)

# 先算：
# False and True -> False
# 再算：
# True or False -> True
````

````python
print(True or False and True)
print(False and False or True)
print(True or False and True and False or True and False or True and True)
print(1 == '1' or 1 >= 1 and False or True)
````

## 6. 綜合練習：字串重複輸出

**範例輸入**
```text
# 5 4 3 2 1
```

限制：
- 只能使用一個 `input()`
- 只能使用一個 `print()`

**範例輸出**
```text
#####
####
###
##
#
```


````python
text, num1, num2, num3, num4, num5 = input().split()
num1, num2, num3, num4, num5 = (
    int(num1),
    int(num2),
    int(num3),
    int(num4),
    int(num5),
)

print(
    text * num1,
    text * num2,
    text * num3,
    text * num4,
    text * num5,
    sep='\n',
)

print(
    text * num1 + '\n'
    + text * num2 + '\n'
    + text * num3 + '\n'
    + text * num4 + '\n'
    + text * num5
)
````

## 8. 綜合練習：秒數換算

**範例輸入**
```text
125
```

目標包含：
- 判斷是否剛好是整分鐘
- 算出分鐘與秒數
- 輸出 `2分5秒`


````python
sec = int(input())

m = sec // 60
s = sec % 60

print(m, s)
print(str(m) + '分' + str(s) + '秒')
print(str(m), '分', str(s), '秒', sep='')
````
