## 1. 字串輸入與輸出格式

先讀取一個名字，再輸出：

```text
Go, 名字, go go
```

使用：

```python
sep=""
```

可以讓 `print()` 的多個輸出項目之間不自動加入空格。


````python
# Go, 人名, go go
# Go, Snail, go go

name = input()
print("Go, ", name, ", go go", sep="")
````

### 使用 `+` 串接字串

相同的輸出也可以使用字串串接完成。


````python
name = input()
print("Go, " + name + ", go go")
````

### 非字串資料要先轉成字串

如果要使用 `+` 把數字放進字串中，可以先使用：

```python
str(...)
```


````python
print("Go, " + str(3) + ", go go")
````

## 2. `input()` 與型別轉換

`input()` 取得的資料是字串。

如果要進行數值運算，可以使用：

```python
int(...)
```

轉成整數。

下面的例子把輸入值轉成整數後，計算它的 30%。


````python
x = input()  # -> 轉型
print(30 / 100 * int(x))
````

## 3. 邏輯運算子 `and` 與 `or`

Boolean 運算常見兩個運算子：

```text
and
or
```

一般來說，`and` 的優先順序高於 `or`。

下面保留原始 notebook 的幾個判斷例子。


````python
# and -> or

print(True or False and True)   # True or False -> True
print(False and False or True)  # False or True -> True

print(
    True or False and True and False
    or True and False
    or True and True
)

# True or False or False or True -> True

print(
    1 == "1" or 1 >= 1 and False or True
)
# False or True and False or True
# -> False or False or True
# -> True
````

## 4. Q1：四則運算

輸入兩個整數 `a`、`b`，依序輸出：

- `a + b`
- `a - b`
- `a * b`
- `a / b`


````python
# Q1
a = int(input())
b = int(input())

print(a + b)
print(a - b)
print(a * b)
print(a / b)
````

## 5. Q2：次方

輸入兩個整數 `a`、`b`，輸出：

```text
a 的 b 次方
```

Python 使用：

```python
a ** b
```

表示次方。


````python
# Q2
a, b = input().split()
a, b = int(a), int(b)

print(a ** b)
````

## 6. Q3：班級分組

輸入學生總人數 `n` 和每班人數 `k`，輸出：

1. 可以分成幾班
2. 剩下幾人

這裡使用：

```text
//  整數除法
%   餘數
```


````python
# Q3
n = int(input())
k = int(input())

print(n // k)
print(n % k)
````

## 10. Q7：字串重複

原始題目文字寫的是：

> 輸入一個字串和一個整數，輸出重複後的字串。

原始程式則先讀取整數，再讀取字串，並使用：

```python
整數 * 字串
```

完成字串重複。

> 這裡保留原始 notebook 的輸入順序與程式邏輯，不另外修改。


````python
a = int(input())
b = input()

c = a * b
print(c)
````

## 11. 練習：總小時數轉成天與小時

輸入總小時數，轉換成：

```text
天 小時
```

### 範例輸入

```text

```

### 範例輸出

```text

```

## 12. 練習：登入檢查

正確帳號：

```text

```

正確密碼：

```text

```

讓使用者同一行輸入帳號與密碼。

## 13. 練習：字串重複圖形
