## 3. 基本資料型態

常見型態：
- `int`：整數
- `float`：浮點數
- `str`：字串
- `bool`：布林值
- `list`：串列


````python
print(6 + 1)    # 整數
print('06')     # 字串
print(5.0)      # 浮點數

# 5 和 5.0 的數值看起來相同，但型態不同。
# int + float 時，運算結果會是 float。
print(5 + 5.0)
````

### 浮點數精確度

有些十進位小數無法用二進位浮點數精確表示，因此計算結果可能出現很小的誤差。


````python
print(0.3 == 0.1 + 0.1 + 0.1)

# 二進位例子：
# 10 = 1 * 2^3 + 0 * 2^2 + 1 * 2^1 + 0 * 2^0
# 0.3 = 3 / 10，但 1/10 無法用有限位二進位小數精確表示。

print(0.1 + 0.1 + 0.1)  # 0.30000000000000004
````

### 布林值 `bool`

布林值只有兩種：
- `True`
- `False`

可以把它想成條件是否成立。


````python
print(10 > 5)  # True
````

### 串列 `list`

串列可以放入多個資料，也可以包含不同型態的資料。


````python
lst = [1, 'ArithmeticError', 3.14, True, [1, 2, 3]]

print(lst[0])
print(lst)
````

### 使用 `type()` 查看資料型態


````python
name = '小明'
num = 10
fl = 3.14
bol = True
lst = [1, 2, 3]

print(type(name))  # <class 'str'>
print(type(num))   # <class 'int'>
print(type(fl))    # <class 'float'>
print(type(bol))   # <class 'bool'>
print(type(lst))   # <class 'list'>
````

## 5. 練習：建立 `intro` 串列

建立一個串列 `intro`，放入：
1. 姓名
2. 年齡
3. 是否喜歡 Python

然後印出 `intro`。

**範例輸入**
```text
John 19 True
```


````python
name, age, like = input().split()

age = int(age)
like = bool(like)  # 將字串轉換為布林值

intro = [name, age, like]
print(intro)
````

### 注意：不要覆蓋內建函式名稱

例如把 `print` 當作變數名稱，之後原本的 `print()` 就會被覆蓋。


````python
# print = 12

# print(12)
# 原本的 print() 是內建函式。
# 如果用變數取代 print，之後 print() 就會失效。
````

### `bool()` 的基本行為


````python
print(bool('False'))  # 非空字串 -> True
print(bool(''))       # 空字串 -> False

print(bool([]))       # 空串列 -> False
print(bool([2, 3]))   # 非空串列 -> True
print(bool([[]]))     # 最外層串列有內容 -> True
````
