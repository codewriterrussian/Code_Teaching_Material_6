## 1. 一行輸入姓名與年齡

讓使用者同一行輸入：

```text
姓名 年齡
```

### 範例輸入

```text
Ivy 14
```

### 預期輸出

```text
姓名：Ivy
年齡：14
```

可以使用：

```python
input().split()
```

把同一行的兩個資料拆開。


````python
name, age = input().split()  # a = input("輸入姓名,年齡:")

print(f"姓名：{name}\n年齡：{age}")
````

## 2. 使用一個 `print()` 連接三次輸入

### 預期輸出

## 3. Boolean 與 List

### Boolean

Boolean（布林值）常用來表示判斷結果：

```text
True
False
```

### List

List（串列）可以保存多個資料，而且元素可以是不同型別。

例如：

```python
[1, 2, "ab", True, [1, 2]]
```


````python
lst = [1, 2, "ab", True, [1, 2]]

print(lst)
print(lst[0])
print(lst[1])

# print(lst[5])  # list index out of range
````

### List Index

List 的第一個元素 index 是：

```text
0
```

因此：

```python
lst[0]
```

會取得第一個元素。

如果使用超出範圍的 index，例如原始程式中的：

```python
lst[5]
```

就會產生 `list index out of range`。

## 6. `input()` 與 `input().split()` 的差別

單純使用：

```python
input()
```

會得到一個字串。

使用：

```python
input().split()
```

則會把輸入拆成一個 List。


````python
x = input()
print(x)
print(type(x))

x = input().split()
print(x)
print(type(x))
````

### 觀察型別

可以使用：

```python

```

例如：

```text

→ str

input().split()
→ list
```


## 7. 字串轉成 Boolean

下面的例子觀察：

```python
bool(...)
```

作用在字串時的結果。


````python
print(bool("True"))
print(bool("False"))

print(bool(""))

# 沒有東西的字串才是 False
# 有東西的字串才是 True
````

### 空字串與非空字串

在原始程式中：

```python
bool("True")
bool("False")
```

都會得到 `True`，因為兩者都是**非空字串**。

而：

```python
bool("")
```

會得到 `False`。

重點是判斷字串是否為空，而不是字串內容是否寫著 `"True"` 或 `"False"`。
