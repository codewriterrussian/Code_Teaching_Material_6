## 1. 時間換算與 `% 24`

原始程式：


````python
h = int(input())

h = h - 15
h = h % 24

print(h)
````

### 為什麼最後要 `% 24`？

一天的 24 小時可以看成一個循環：

```text
0, 1, 2, ..., 22, 23
↓
再往下一小時
↓
0
```

所以：

```python
h % 24
```

可以把結果重新限制在：

```text
0 ~ 23
```

例如：

```text
25 % 24 = 1
```

表示超過一天後又回到第 1 小時。

原始註記也特別測試了負數：


````python
-3 % 24
````

在 Python 中：

```text
-3 % 24 = 21
```

所以即使先減去一個較大的時差而得到負數，`% 24` 仍可以把結果拉回一天的時刻範圍。

> 這個負數 `%` 的具體結果是 Python 的行為；若之後改成 C/C++，不要直接假設所有語言對負數餘數的規則完全相同。


## 2. 用公式做溫度轉換

原始程式：


````python
r = float(input())

R = r * 9 / 5 + 32

print(format(R, "g"))
print(format(R, "f"))
````

```text
120.2

```

## 3. 用 `1 - a` 切換 0 與 1

原始程式：


````python
a = int(input())

a = 1 - a

print(a)
````

如果題目中的 `a` 保證只有：

```text
0 或 1
```

那麼：

```python
a = 1 - a
```

可以完成切換：

```text
a = 0
→ 1 - 0
→ 1

a = 1
→ 1 - 1
→ 0
```

也就是：

```text
0 ↔ 1
```

這是一個很簡短的二元狀態切換方式。

> 原始 notebook 沒有附輸入限制，因此「只會輸入 0 或 1」是這個技巧成立所需要的條件，不應從程式本身直接當成已知題目規則。


## 4. `//`：整數分組

原始註記：

```text
1, 2, 3   → 第 1 組
4, 5, 6   → 第 2 組
7, 8, 9   → 第 3 組
...
```

程式：


````python
n = int(input())
n -= 1

print(n // 3 + 1)
````

### 為什麼先 `n -= 1`？

假設：

```text
1, 2, 3
```

都要得到：

```text
1
```

先減 1：

```text
1, 2, 3
↓
0, 1, 2
```

再做：

```text
0 // 3 = 0
1 // 3 = 0
2 // 3 = 0
```

最後加 1：

```text
1, 1, 1
```

下一組：

```text
4, 5, 6
↓ 減 1
3, 4, 5
↓ // 3
1, 1, 1
↓ + 1
2, 2, 2
```

所以：

```python
(n - 1) // 3 + 1
```

可以把連續編號每 3 個分成一組。


## 5. `//` 與 `%` 一起使用：整組 + 剩下

原始程式：


````python
n = int(input())

dozens = n // 12
single = n % 12

price = dozens * 50 + single * 5

print(price)
````

這裡把數量拆成：

```text
完整的 12 個一組
+
剩下不到 12 個
```

因此：

```python
dozens = n // 12
```

得到有幾個完整的 12。

而：

```python
single = n % 12
```

得到剩下幾個。

可以記成：

```text
// → 有幾整組
%  → 剩多少
```

這兩個運算子常常一起使用。


## 6. BMI 與 f-string 格式化

原始程式：


````python
weight = int(input())
height = float(input())

bmi = weight / (height * height)

print(f"{bmi:.1f}")
````

計算：

```text
BMI = weight / height²
```

最後：

```python
f"{bmi:.1f}"
```

表示：

> 用浮點數格式輸出，並保留小數點後 1 位。

例如：

```text
23.456
```

會顯示成類似：

```text
23.5
```


## 7. `%f` 浮點數格式

原 notebook 特別留下標題：

```text
%f
```

並測試：


````python
print("%.10f" % 5)
print("%6.3f" % 5)
print("%2.3f" % 5)
````

原 notebook 的輸出：

```text
5.0000000000
 5.000
5.000
```

### `%.10f`

```text
小數點後 10 位
```

所以：

```text
5 → 5.0000000000
```

### `%6.3f`

可以先把它拆成：

```text
6 → 最小欄位寬度
3 → 小數點後 3 位
f → 浮點數格式
```

因此 `5` 顯示為：

```text
 5.000
```

前面補一個空格，使整體寬度至少為 6。

### `%2.3f`

雖然指定欄位寬度 `2`，但：

```text
5.000
```

本身已經超過 2 個字元，所以不會把數字截短。

> 原 notebook 在這裡使用的是 Python 的舊式 `%` 格式化語法。它和 C/C++ `printf` 的格式字串長得很像，因此很適合拿來建立格式概念，但兩者仍是不同語言環境。


## 8. 函式：輸入與輸出

原始函式：


````python
def out(a, b):
    return a + b
````

這個函式可以拆成：

```text
函式名稱：out
輸入：a、b
輸出：a + b
```

呼叫時需要提供兩個引數，例如：

```python
out(3, 5)
```

會回傳：

```text
8
```

### 原 notebook 中的錯誤示範

原始筆記接著寫：

```python
a = out()
```

但 `out` 定義時需要：

```text
a
b
```

兩個參數，因此沒有提供引數時不能正常呼叫。

這可以用來提醒：

> **函式呼叫時提供的引數數量，要符合函式定義需要的參數。**


## 9. `print()`：有動作，但沒有一般資料回傳值

原 notebook 留下註記：

```text
print -> 有輸入、執行、空輸出
```

並觀察到 `None`。

為了避免依賴前面 notebook 的舊變數狀態，可以把示範整理成：


````python
a = "a"

b = print(a)
print(b)
````

執行流程：

```text
b = print(a)
        ↓
先把 a 印到螢幕
        ↓
print() 本身沒有回傳一般資料值
        ↓
b 得到 None
```

所以輸出會是：

```text
a
None
```

重要差別：

```text
印出一個值
≠
return 一個值
```

`print()` 的主要工作是輸出到螢幕；它的回傳值是 `None`。


## 10. `enumerate()`：同時取得索引與元素

原始程式：


````python
lst = ['1', '2', 3, 5, ['a']]

for idx, ele in enumerate(lst):
    print(idx, ele)
````

`enumerate(lst)` 每一次會提供兩個東西：

```text
idx → 索引
ele → 該位置的元素
```

因此可以理解成：

```text
索引 0 → '1'
索引 1 → '2'
索引 2 → 3
索引 3 → 5
索引 4 → ['a']
```

原 notebook 的輸出為：

```text
0 1
1 2
2 3
3 5
4 ['a']
```

注意這個串列故意混合了不同型態：

```python
'1'
'2'
3
5
['a']
```

`enumerate()` 仍然只是依序提供索引和值，不要求所有元素都必須是同一型態。
